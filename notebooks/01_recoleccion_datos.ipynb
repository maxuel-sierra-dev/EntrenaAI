# 01 - Recolección de datos

## Objetivo

Obtener datos históricos de partidos de fútbol americano universitario (College Football) mediante la API de CollegeFootballData.

Se recopilarán datos de los encuentros, incluyendo fechas, equipos participantes y resultados, para construir posteriormente un dataset que permita entrenar modelos de Machine Learning y predecir probabilidades de victoria.

```python
import pandas as pd
import requests

from getpass import getpass
```

```python
API_URL = "https://api.collegefootballdata.com"

API_KEY = getpass("Introduce tu API Key: ")

headers = {
    "Authorization": f"Bearer {API_KEY}"
}
```

```python
params = {
    "year": 2024
}

response = requests.get(
    f"{API_URL}/games",
    headers=headers,
    params=params,
    timeout=30
)

print("Código de respuesta:", response.status_code)
```

```python
games = response.json()

df_games = pd.DataFrame(games)

print("Cantidad de partidos:", len(df_games))
print("Cantidad de columnas:", len(df_games.columns))

display(df_games.head())
```

```python
print("Columnas disponibles:\n")

for columna in df_games.columns:
    print(columna)
```

```python
df_games.info()
```

```python
print("Valores nulos por columna:")
display(df_games.isnull().sum())
```

```python
df_games.to_csv(
    "../data/partidos_2024.csv",
    index=False,
    encoding="utf-8"
)

print("Dataset guardado correctamente.")
```
